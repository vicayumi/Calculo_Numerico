**Exercício 1 — Sistemas na forma matricial $A\mathbf{x}=\mathbf{b}$**

a)
$$
\begin{cases} 2x + y = 7 \\ x - 3y = -5 \end{cases}
\quad\Longrightarrow\quad
\underbrace{\begin{bmatrix} 2 & 1 \\ 1 & -3 \end{bmatrix}}_{A}
\underbrace{\begin{bmatrix} x \\ y \end{bmatrix}}_{\mathbf{x}}
=
\underbrace{\begin{bmatrix} 7 \\ -5 \end{bmatrix}}_{\mathbf{b}}
$$

- **Matriz dos coeficientes:** $A = \begin{bmatrix} 2 & 1 \\ 1 & -3 \end{bmatrix}$
- **Vetor de incógnitas:** $\mathbf{x} = \begin{bmatrix} x \\ y \end{bmatrix}$
- **Vetor independente:** $\mathbf{b} = \begin{bmatrix} 7 \\ -5 \end{bmatrix}$

### b)
$$
\begin{cases} x + 2y - z = 2 \\ 3x - y + 2z = 9 \\ 2x + y + z = 7 \end{cases}
\quad\Longrightarrow\quad
\underbrace{\begin{bmatrix} 1 & 2 & -1 \\ 3 & -1 & 2 \\ 2 & 1 & 1 \end{bmatrix}}_{A}
\underbrace{\begin{bmatrix} x \\ y \\ z \end{bmatrix}}_{\mathbf{x}}
=
\underbrace{\begin{bmatrix} 2 \\ 9 \\ 7 \end{bmatrix}}_{\mathbf{b}}
$$

- **Matriz dos coeficientes:** $A = \begin{bmatrix} 1 & 2 & -1 \\ 3 & -1 & 2 \\ 2 & 1 & 1 \end{bmatrix}$
- **Vetor de incógnitas:** $\mathbf{x} = \begin{bmatrix} x \\ y \\ z \end{bmatrix}$
- **Vetor independente:** $\mathbf{b} = \begin{bmatrix} 2 \\ 9 \\ 7 \end{bmatrix}$

*(Verificação numérica na célula abaixo.)*

```python
# ex1 — verificação numérica das formas matriciais

# a)
A1a = np.array([[2, 1],
                [1, -3]], dtype=float)
b1a = np.array([7, -5], dtype=float)
print("1a) solução:", np.linalg.solve(A1a, b1a), "  (esperado: x=16/7, y=17/7)")
print("    16/7 =", 16/7, "| 17/7 =", 17/7)

# b)
A1b = np.array([[1, 2, -1],
                [3, -1, 2],
                [2, 1, 1]], dtype=float)
b1b = np.array([2, 9, 7], dtype=float)
print("1b) solução:", np.linalg.solve(A1b, b1b), "  (esperado: x=2, y=1, z=2)")
```

**Interpretação:** escrever o sistema como $A\mathbf{x}=\mathbf{b}$ separa a estrutura (coeficientes em $A$) dos dados ($\mathbf{b}$), o que permite aplicar algoritmos gerais (como Gauss) a qualquer sistema. Os resultados confirmam: (a) $x = 16/7,\ y = 17/7$; (b) $x=2,\ y=1,\ z=2$.


**Exercício 2 — Eliminação de Gauss (manual)**

$$
\begin{cases} 2x + y - z = 8 \\ -3x - y + 2z = -11 \\ -2x + y + 2z = -3 \end{cases}
$$

Matriz aumentada inicial:
$$
\left[\begin{array}{ccc|c} 2 & 1 & -1 & 8 \\ -3 & -1 & 2 & -11 \\ -2 & 1 & 2 & -3 \end{array}\right]
$$

Passo 1 — zerar abaixo do pivô $a_{11}=2$:
multiplicadores $m_{21} = \frac{-3}{2} = -\frac32$ e $m_{31} = \frac{-2}{2} = -1$.

- $L_2 \leftarrow L_2 - m_{21}L_1 = L_2 + \tfrac32 L_1$
- $L_3 \leftarrow L_3 - m_{31}L_1 = L_3 + L_1$

$$
\left[\begin{array}{ccc|c} 2 & 1 & -1 & 8 \\ 0 & \tfrac12 & \tfrac12 & 1 \\ 0 & 2 & 1 & 5 \end{array}\right]
$$

Passo 2 — zerar abaixo do pivô $a_{22}=\tfrac12$:
multiplicador $m_{32} = \frac{2}{1/2} = 4$.

- $L_3 \leftarrow L_3 - 4L_2$

$$
\left[\begin{array}{ccc|c} 2 & 1 & -1 & 8 \\ 0 & \tfrac12 & \tfrac12 & 1 \\ 0 & 0 & -1 & 1 \end{array}\right]
$$

Substituição regressiva:

- $-z = 1 \Rightarrow z = -1$
- $\tfrac12 y + \tfrac12 z = 1 \Rightarrow \tfrac12 y - \tfrac12 = 1 \Rightarrow y = 3$
- $2x + y - z = 8 \Rightarrow 2x + 3 + 1 = 8 \Rightarrow x = 2$

$$\boxed{x = 2,\quad y = 3,\quad z = -1}$$

Verificação: $2(2)+3-(-1)=8$ | $-3(2)-3+2(-1)=-11$ | $-2(2)+3+2(-1)=-3$

# Parte 2 — Propriedades e interpretação

**Exercício 3 — Número de soluções:**

a)
$\begin{cases} x + y = 3 \\ 2x + 2y = 6 \end{cases}$

$$
\left[\begin{array}{cc|c} 1 & 1 & 3 \\ 2 & 2 & 6 \end{array}\right]
\xrightarrow{L_2 \leftarrow L_2 - 2L_1}
\left[\begin{array}{cc|c} 1 & 1 & 3 \\ 0 & 0 & 0 \end{array}\right]
$$

A segunda linha vira $0=0$ (sempre verdadeira): há 1 equação independente para 2 incógnitas. Portanto infinitas soluções: $y = t$ livre, $x = 3 - t$, com $t\in\mathbb{R}$.

b)
$\begin{cases} x + y = 3 \\ 2x + 2y = 7 \end{cases}$

$$
\left[\begin{array}{cc|c} 1 & 1 & 3 \\ 2 & 2 & 7 \end{array}\right]
\xrightarrow{L_2 \leftarrow L_2 - 2L_1}
\left[\begin{array}{cc|c} 1 & 1 & 3 \\ 0 & 0 & 1 \end{array}\right]
$$

A segunda linha vira $0 = 1$ (contradição). Portanto nenhuma solução (sistema incompatível).

c)
$\begin{cases} x + y = 3 \\ 2x - y = 0 \end{cases}$

$$
\left[\begin{array}{cc|c} 1 & 1 & 3 \\ 2 & -1 & 0 \end{array}\right]
\xrightarrow{L_2 \leftarrow L_2 - 2L_1}
\left[\begin{array}{cc|c} 1 & 1 & 3 \\ 0 & -3 & -6 \end{array}\right]
$$

Há dois pivôs (um por coluna de incógnita). Então $-3y=-6 \Rightarrow y=2$ e $x = 3 - 2 = 1$. Solução única: $(x,y)=(1,2)$.

### Resumo (critério pelo posto)
| Sistema | posto($A$) | posto($[A\,|\,b]$) | nº de incógnitas | Conclusão |
|---|---|---|---|---|
| a) | 1 | 1 | 2 | infinitas soluções |
| b) | 1 | 2 | 2 | nenhuma solução |
| c) | 2 | 2 | 2 | solução única |

```python
# ex3 — verificação com o posto (rank) das matrizes
sistemas = {
    "a)": (np.array([[1, 1], [2, 2]], float), np.array([3, 6], float)),
    "b)": (np.array([[1, 1], [2, 2]], float), np.array([3, 7], float)),
    "c)": (np.array([[1, 1], [2, -1]], float), np.array([3, 0], float)),
}

for nome, (A, b) in sistemas.items():
    Ab = np.column_stack((A, b))              # matriz aumentada [A | b]
    rA = np.linalg.matrix_rank(A)
    rAb = np.linalg.matrix_rank(Ab)
    n = A.shape[1]

    if rA < rAb:
        res = "NENHUMA solução (incompatível)"
    elif rA == n:
        res = "solução ÚNICA"
    else:
        res = "INFINITAS soluções"
    print(f"{nome} posto(A)={rA}, posto([A|b])={rAb}, incógnitas={n} -> {res}")
```

**Interpretação:** pelo Teorema de Rouché–Capelli, o sistema é compatível se e somente se $\text{posto}(A)=\text{posto}([A|b])$. Sendo compatível, a solução é única quando o posto é igual ao número de incógnitas, e há infinitas soluções quando é menor. A redução da matriz aumentada mostra isso diretamente: linha nula $\to$ infinitas; linha $0=c\neq0$ $\to$ nenhuma; pivô em todas as colunas $\to$ única.

# Parte 3 — Implementação numérica em Python

**Exercício 4 — Eliminação de Gauss sem pivoteamento:**

```python
def gauss_sem_pivoteamento(A, b, verbose=False):
    '''
    Resolve A x = b por eliminação de Gauss SEM pivoteamento.

    Parâmetros
    ----------
    A : array (n x n)  matriz dos coeficientes
    b : array (n,)     vetor independente
    verbose : bool     se True, imprime a matriz aumentada a cada etapa

    Retorna
    -------
    x : array (n,)     solução calculada
    '''
    A = np.array(A, dtype=float)
    b = np.array(b, dtype=float)
    n = len(b)

    # 1) Forma a matriz aumentada [A | b]
    Ab = np.column_stack((A, b))
    if verbose:
        print("Matriz aumentada inicial:\n", Ab, "\n")

    # 2) Eliminação: zera os elementos abaixo de cada pivô
    for k in range(n - 1):                     # k = coluna do pivô
        pivo = Ab[k, k]
        if abs(pivo) < 1e-12:                  # sem pivoteamento: pivô nulo = falha
            raise ZeroDivisionError(
                f"Pivô nulo (ou quase) na posição ({k},{k}). "
                "Sem pivoteamento o método falha."
            )
        for i in range(k + 1, n):              # i = linhas abaixo do pivô
            m = Ab[i, k] / pivo                # multiplicador m_ik
            Ab[i, k:] = Ab[i, k:] - m * Ab[k, k:]   # L_i <- L_i - m * L_k
            if verbose:
                print(f"L{i+1} <- L{i+1} - ({m:.4f}) * L{k+1}")
        if verbose:
            print(f"\nApós eliminar a coluna {k+1}:\n", Ab, "\n")

    # Verifica o último pivô
    if abs(Ab[n - 1, n - 1]) < 1e-12:
        raise ZeroDivisionError("Último pivô nulo: sistema singular.")

    # 3) Substituição regressiva (de baixo para cima)
    x = np.zeros(n)
    for i in range(n - 1, -1, -1):
        soma = np.dot(Ab[i, i + 1:n], x[i + 1:n])   # termos já conhecidos
        x[i] = (Ab[i, n] - soma) / Ab[i, i]

    return x
```

```python
# Teste com o sistema do exercício 2
A2 = np.array([[ 2,  1, -1],
               [-3, -1,  2],
               [-2,  1,  2]], dtype=float)
b2 = np.array([8, -11, -3], dtype=float)

x_gauss = gauss_sem_pivoteamento(A2, b2, verbose=True)
x_numpy = np.linalg.solve(A2, b2)

print("Solução (Gauss implementado):", x_gauss)
print("Solução (numpy.linalg.solve):", x_numpy)
print("Diferença absoluta máxima   :", np.max(np.abs(x_gauss - x_numpy)))
print("Resíduo ||Ax - b||          :", np.linalg.norm(A2 @ x_gauss - b2))
```

```python
# Limitação do método sem pivoteamento: pivô nulo
A_ruim = np.array([[0, 1],
                   [1, 1]], dtype=float)   # tem solução, mas a11 = 0
b_ruim = np.array([1, 2], dtype=float)

try:
    gauss_sem_pivoteamento(A_ruim, b_ruim)
except ZeroDivisionError as e:
    print("Erro esperado:", e)

print("numpy.linalg.solve (usa pivoteamento):", np.linalg.solve(A_ruim, b_ruim))
```

**Interpretação:** o código reproduz exatamente os passos manuais do Exercício 2 (multiplicadores $-3/2$, $-1$ e $4$) e obtém $x=2,\ y=3,\ z=-1$. A diferença para numpy.linalg.solve é da ordem do erro de arredondamento (≈ 0), e o resíduo $\|A\mathbf{x}-\mathbf{b}\|$ é praticamente nulo.

A última célula mostra a limitação do método sem pivoteamento: se algum pivô for zero (ou muito pequeno), ocorre divisão por zero (ou amplificação de erros de arredondamento), mesmo que o sistema tenha solução. O numpy.linalg.solve evita isso usando fatoração LU com pivoteamento parcial.

**Exercício 5 — Modelo de balanço térmico:**

$$
\begin{bmatrix} 4 & -1 & 0 \\ -1 & 4 & -1 \\ 0 & -1 & 3 \end{bmatrix}
\begin{bmatrix} T_1 \\ T_2 \\ T_3 \end{bmatrix}
=
\begin{bmatrix} 15 \\ 10 \\ 10 \end{bmatrix}
$$

a) Resolução manual (Gauss):

Matriz aumentada:
$$
\left[\begin{array}{ccc|c} 4 & -1 & 0 & 15 \\ -1 & 4 & -1 & 10 \\ 0 & -1 & 3 & 10 \end{array}\right]
$$

Passo 1 — pivô $a_{11}=4$; $m_{21} = -\tfrac14$, $m_{31}=0$:
$L_2 \leftarrow L_2 + \tfrac14 L_1$

$$
\left[\begin{array}{ccc|c} 4 & -1 & 0 & 15 \\ 0 & \tfrac{15}{4} & -1 & \tfrac{55}{4} \\ 0 & -1 & 3 & 10 \end{array}\right]
$$

Passo 2 — pivô $a_{22}=\tfrac{15}{4}$; $m_{32} = \dfrac{-1}{15/4} = -\tfrac{4}{15}$:
$L_3 \leftarrow L_3 + \tfrac{4}{15} L_2$

$$
\left[\begin{array}{ccc|c} 4 & -1 & 0 & 15 \\ 0 & \tfrac{15}{4} & -1 & \tfrac{55}{4} \\ 0 & 0 & \tfrac{41}{15} & \tfrac{41}{3} \end{array}\right]
$$

Substituição regressiva:

- $\tfrac{41}{15}T_3 = \tfrac{41}{3} \Rightarrow T_3 = 5$
- $\tfrac{15}{4}T_2 - T_3 = \tfrac{55}{4} \Rightarrow \tfrac{15}{4}T_2 = \tfrac{75}{4} \Rightarrow T_2 = 5$
- $4T_1 - T_2 = 15 \Rightarrow 4T_1 = 20 \Rightarrow T_1 = 5$

$$\boxed{T_1 = T_2 = T_3 = 5}$$

Verificação: $4(5)-5=15$ ✓ | $-5+4(5)-5=10$ ✓ | $-5+3(5)=10$ ✓

b) Implementação em Python:

```python
# ex 5b — resolve o balanço térmico reutilizando a função do exercício 4
A5 = np.array([[ 4, -1,  0],
               [-1,  4, -1],
               [ 0, -1,  3]], dtype=float)
b5 = np.array([15, 10, 10], dtype=float)

T = gauss_sem_pivoteamento(A5, b5, verbose=True)
T_np = np.linalg.solve(A5, b5)

print("Temperaturas (Gauss)       : T1 = %.4f, T2 = %.4f, T3 = %.4f" % tuple(T))
print("Temperaturas (numpy solve) :", T_np)
print("Diferença máxima           :", np.max(np.abs(T - T_np)))
print("Resíduo ||AT - b||         :", np.linalg.norm(A5 @ T - b5))
```

**Interpretação:** as três temperaturas de equilíbrio são iguais a $5$ (na unidade do problema). A matriz é diagonalmente dominante (cada $|a_{ii}|$ é maior que a soma dos demais elementos da linha: $4>1$, $4>2$, $3>1$), o que garante pivôs não nulos e boa estabilidade numérica — por isso a eliminação sem pivoteamento funciona bem aqui. O resultado manual coincide com o Python e com o numpy, e o resíduo é praticamente zero.
